```python
import matplotlib.pyplot as plt #plt work as alias/nicname
#pyplot is a collection of functions
```

```python
#data
x = [1, 2, 3, 4, 5]
y = [10, 20, 35, 40, 50]

plt.figure(figsize=[4,3]) # to fix the size of the graph
plt.plot(x, y)
plt.show()
```

```python
#customised chat
plt.figure(figsize=[4,4])
plt.plot(x,y, color='g', marker='*', linestyle='dashed', linewidth= 1, markersize=12)
plt.title("Sales Data") # title of the chart
plt.xlabel("x-axis") # title of x- axis
plt.ylabel("y-axis") # title of y- axis
plt.show()
```

ADVANCED MULTILINE CART & LEGEND

```python
#data
plt.figure(figsize=[4,4])
x = [2010, 2015, 2020, 2025, 2030]
plt.title("Sales Report")
y1 = [10, 20, 30, 45, 50]
y2 = [12, 22, 38, 48, 60]
plt.plot(x, y1, label= "textile")
plt.plot(x, y2, label= "furniture")
plt.xlabel("Years") # title of x- axis
plt.ylabel("Sales in a year") # title of y- axis
plt.legend() # defines the lines
plt.show()
```

```python
# data
x = [2010,2015,2020,2025,2030]
y1 = [10,20,35,45,50]
y2 = [12,22,38,48,60]
 
plt.figure(figsize=(4,4))
 
plt.plot(x,y1,label="textile",color='y', marker='v', linestyle=':',linewidth=3, markersize=12)
plt.plot(x,y2,label="furniture",color='b', marker='v', linestyle=':',linewidth=3, markersize=12)
 
plt.title("Sales report") # title top show
plt.xlabel("Years") # x-cor | years
plt.ylabel("Sales") # y-cor | sales
 
plt.legend() # legent define the lines
 
plt.show()
```

```python
plt.figure(figsize=(4,4))
x = ['a','b','c','d','e']
y = [10, 20, 30, 40, 50]
plt.bar(x,y)
plt.title("Data Reports") # title of the chart
plt.xlabel("x-axis") # title of x- axis
plt.ylabel("y-axis")
plt.show()
```

Customise Bar chart

```python
plt.figure(figsize=(4,4))
x = ['a','b','c','d','e']
y = [10, 20, 30, 40, 50]
plt.bar(x,y, color= ['orange','blue','green', 'red', 'black'], edgecolor='purple',
       linewidth=2)
plt.title("Data Reports") # title of the chart
plt.xlabel("x-axis") # title of x- axis
plt.ylabel("y-axis")
plt.show()
```

```python
# data
x = [2010,2015,2020,2025,2030]
y1 = [10,15,35,45,50]
y2 = [20,22,38,48,60]
y3 = [30,45,52,66,75]
 
plt.figure(figsize=(4,4))
 
plt.plot(x,y1,label="textile",color='y', marker='v',linewidth=3, markersize=12)
plt.plot(x,y2,label="furniture",color='b', marker='v',linewidth=3, markersize=12)
plt.plot(x,y3,label="decor",color='g', marker='v',linewidth=3, markersize=12)

plt.title("Sales report") # title top show
plt.xlabel("Years") # x-cor | years
plt.ylabel("Sales") # y-cor | sales
 
plt.legend() # legent define the lines
 
plt.show()
```

```python
#multibar chart

import matplotlib.pyplot as plt
import numpy as np

plt.figure(figsize=(6,5))
x = np.array(['Category 1','Category 2','Category 3','Category 4','Category 5'])
y1 = [50, 100, 150, 200, 250]
y2 = [60, 120, 180, 240, 300]
y3 = [100, 200, 300, 400, 500]

x_pos = np.arange(len(x))
#print(x_pos)

w_width = 0.30
plt.bar(x_pos - w_width, y1, label="girls", color= 'yellow', width = w_width)
plt.bar(x_pos, y2, label="boys", color='green', width = w_width)
plt.bar(x_pos  +  w_width, y3, label="trans", color= 'blue', width = w_width)

plt.title("Grouped Bar Plot") # title of the chart
plt.xlabel("Category") # title of x- axis
plt.ylabel("Values")
plt.legend()
plt.grid(True)
plt.xticks(x_pos,x)
plt.show()
```

Histogram Chart

```python
import matplotlib.pyplot as plt
data = [2,3,4,5,6,7,8,9,12,23,24,45,13] #rating data

plt.figure(figsize=(4,4))
plt.hist(data, bins=3) #bad, average, good
plt.title("Histogram")
plt.xlabel("Data") # title of x- axis
plt.ylabel("Frequency")
plt.show()
```

SubPlot

```python
#data -1 -> bar chart
categories = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri']
sales = [10, 20, 30, 40, 50]

#data 2 -> scatter Plot
x = [2000, 2010, 2020, 2030, 2040]
y = [20, 40, 60, 80, 100]

plt.figure(figsize=(6,5))
#first plot bar cart
plt.subplot(2,2,1) #row, column 1
plt.bar(categories, sales)
plt.title("Chart box 1")

#second plot -> scatterb plot
plt.subplot(2,2,2) #row, column 2
plt.scatter(x, y)
plt.title("Chart box 2")

#second plot -> scatterb plot
plt.subplot(2,2,3) #row, column 3
plt.scatter(x, y)
plt.title("Chart box 3")

#second plot -> scatterb plot
plt.subplot(2,2,4) #row, column 4
plt.bar(x, y)
plt.title("Chart box 4")

plt.tight_layout()
plt.savefig("subplot.png")
plt.show()
```
