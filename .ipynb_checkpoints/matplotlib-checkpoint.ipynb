```python
import matplotlib.pyplot as plt #plt work as alias/nicname
#pyplot is a collection of functions
```

```python
#data
x = [1, 2, 3, 4, 5]
y = [10, 20, 35, 40, 50]

plt.figure(figsize=[4,3]) # to fix the size of the graph
plt.plot(x, y)
plt.show()
```

```python
#customised chat
plt.figure(figsize=[4,4])
plt.plot(x,y, color='g', marker='*', linestyle='dashed', linewidth= 1, markersize=12)
plt.title("Sales Data") # title of the chart
plt.xlabel("x-axis") # title of x- axis
plt.ylabel("y-axis") # title of y- axis
plt.show()
```

ADVANCED MULTILINE CART & LEGEND

```python
#data
plt.figure(figsize=[4,4])
x = [2010, 2015, 2020, 2025, 2030]
plt.title("Sales Report")
y1 = [10, 20, 30, 45, 50]
y2 = [12, 22, 38, 48, 60]
plt.plot(x, y1, label= "textile")
plt.plot(x, y2, label= "furniture")
plt.xlabel("Years") # title of x- axis
plt.ylabel("Sales in a year") # title of y- axis
plt.legend() # defines the lines
plt.show()
```

```python
# data
x = [2010,2015,2020,2025,2030]
y1 = [10,20,35,45,50]
y2 = [12,22,38,48,60]
 
plt.figure(figsize=(4,4))
 
plt.plot(x,y1,label="textile",color='y', marker='v', linestyle=':',linewidth=3, markersize=12)
plt.plot(x,y2,label="furniture",color='b', marker='v', linestyle=':',linewidth=3, markersize=12)
 
plt.title("Sales report") # title top show
plt.xlabel("Years") # x-cor | years
plt.ylabel("Sales") # y-cor | sales
 
plt.legend() # legent define the lines
 
plt.show()
```

```python
plt.figure(figsize=(4,4))
x = ['a','b','c','d','e']
y = [10, 20, 30, 40, 50]
plt.bar(x,y)
plt.title("Data Reports") # title of the chart
plt.xlabel("x-axis") # title of x- axis
plt.ylabel("y-axis")
plt.show()
```

Customise Bar chart

```python
plt.figure(figsize=(4,4))
x = ['a','b','c','d','e']
y = [10, 20, 30, 40, 50]
plt.bar(x,y, color= ['orange','blue','green', 'red', 'black'], edgecolor='purple',
       linewidth=2)
plt.title("Data Reports") # title of the chart
plt.xlabel("x-axis") # title of x- axis
plt.ylabel("y-axis")
plt.show()
```

```python
# data
x = [2010,2015,2020,2025,2030]
y1 = [10,15,35,45,50]
y2 = [20,22,38,48,60]
y3 = [30,45,52,66,75]
 
plt.figure(figsize=(4,4))
 
plt.plot(x,y1,label="textile",color='y', marker='v',linewidth=3, markersize=12)
plt.plot(x,y2,label="furniture",color='b', marker='v',linewidth=3, markersize=12)
plt.plot(x,y3,label="decor",color='g', marker='v',linewidth=3, markersize=12)

plt.title("Sales report") # title top show
plt.xlabel("Years") # x-cor | years
plt.ylabel("Sales") # y-cor | sales
 
plt.legend() # legent define the lines
 
plt.show()
```

```python
plt.figure(figsize=(4,4))
y = ['a','b','c','d','e']
x1 = [50, 100, 150, 200, 250]
x2 = [6, 12, 18, 24, 30]
x3 = [100, 200, 300, 400, 500]

plt.bar(x1,y, color= 'yellow', edgecolor='purple')
plt.bar(x2,y, color='green', edgecolor='yellow')
plt.bar(x3,y, color= 'blue', edgecolor='green')
plt.title("Data Reports") # title of the chart
plt.xlabel("x-axis") # title of x- axis
plt.ylabel("y-axis")
plt.show()
```
