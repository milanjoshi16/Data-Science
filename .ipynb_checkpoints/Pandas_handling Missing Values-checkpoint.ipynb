```python
import pandas as pd
url = "https://raw.githubusercontent.com/rajendra0968jangid/Ds-Arya/main/file2.json"
df = pd.read_json(url)
print(df)
```

```python
#isnull()
print(df.isnull())
```

```python
#isnull() with sum()
print(df.isnull().sum()) #o/p is rowwise.
```

```python
#isnull() with sum() --> 1-Rows, 0- column for axis (means opposite of numpy. Numpy is just a matrix, but pandas is a table/ column feature sets)
print(df.isnull().sum(axis = 1)) #o/p is columnwise.
```

```python
#drop values by row i.e. rows with null values have been dropped but no changes in data frame.
print(df.dropna())
```

```python
#drop Null values 
print(df.dropna(axis = 1))
```

```python
#fill
print(df.fillna(0))
```

```python
#fill with 100
print(df.fillna(100))
```

```python
import pandas as pd
url = "https://raw.githubusercontent.com/rajendra0968jangid/Ds-Arya/main/file2.json"
df = pd.read_json(url)
print(df)
```

```python
print(df.isnull().sum())
```

```python
print(df.isnull().sum(axis = 1))
```

```python
print(df.dropna())
```

```python
print(df.dropna(axis = 1))
```

```python
print(df.fillna(100))
```

```python
#specific number(position) 
# fill roll no: 300 | marks: 400
df["roll no"] = df["roll no"].fillna(300)
df["marks"] = df["marks"].fillna(400)
print(df)
```

```python
#alternative option 
df.fillna({"roll no": 200, "marks":500}, inplace = True) # no changes because due to above code there is no null value
print(df)
```

```python
#alternative option 
df = pd.read_json(url)
df.fillna({"roll no": 200, "marks":500}, inplace = True) # operation on raw data
print(df)
```

```python
#fill the mean val at NaN 
df = pd.read_json(url)
mean = df["marks"].mean()
print(mean)

df["makrs"] = df["marks"].fillna(mean)
print(df)
```
