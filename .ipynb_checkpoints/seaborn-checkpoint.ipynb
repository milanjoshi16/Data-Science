```python
import seaborn as sns
```

```python
#relation
#bivariate- works on 2 variables
#scatter plot
#lineplot
#relplot- lineplot + scatter plot
tips = sns.load_dataset('tips')
print(tips)
```

```python
print(tips.head()) #print 5 entries from start
```

```python
#scatter plot -> axix level function
tips = tips.rename(columns={"sex":"gender"})
sns.scatterplot(data = tips, x='total_bill',y='tip', hue='gender')
```

```python
sns.relplot(data = tips, x='total_bill',y='tip', hue='gender')
```

```python
sns.lineplot(data = tips, x='total_bill',y='tip', hue='gender')
```

```python
sns.relplot(data = tips, x='total_bill',y='tip', hue='gender', kind = 'line')
```

```python
sns.relplot(data = tips, x='total_bill',y='tip', hue='gender', kind = 'scatter', style= 'time')
```

```python
sns.relplot(data = tips, x='total_bill',y='tip',col='gender', kind = 'line')
```

```python
sns.relplot(data = tips, x='total_bill',y='tip',row='gender', kind = 'line')
```

```python
sns.relplot(data = tips, x='total_bill',y='tip',col='gender', kind = 'scatter', row='time')
```

```python
sns.relplot(data = tips, x='total_bill',y='tip',col='day', kind = 'scatter') #doesn't look presentable so we use column wrap
```

```python
sns.relplot(data = tips, x='total_bill',y='tip',col='day', kind = 'scatter', col_wrap=2)
```
