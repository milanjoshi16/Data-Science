```python
#Find sum of first N natural numbers
n = int(input("Enter a number: "))
i = 1
total = 0
while i <= n:
    total = total + i
    i = i + 1
print("Sum =", total)
```

```python
#Print multiplication table of a number
n = int(input("Enter a number: "))
i = 1
while i <= 10:
    print(n, "x", i, "=", n * i)
    i = i + 1
```

```python
#Count digits of a number
n = int(input("Enter a number: "))
count = 0
while n > 0:
    n = n // 10
    count = count + 1
print("Digits =", count)
```

```python
#Reverse a number
n = int(input("Enter a number: "))
reverse = 0
while n > 0:
    digit = n % 10
    reverse = reverse * 10 + digit
    n = n // 10
print("Reverse =", reverse)
```

```python
#Check if number is palindrome
n = int(input("Enter a number: "))
original = n
reverse = 0
while n > 0:
    digit = n % 10
    reverse = reverse * 10 + digit
    n = n // 10
if original == reverse:
    print("Palindrome")
else:
    print("Not Palindrome")
```

```python
#Find sum of digits
n = int(input("Enter a number: "))
total = 0
while n > 0:
    digit = n % 10
    total = total + digit
    n = n // 10
print("Sum of digits =", total)
```

```python
#Find factorial of a number
n = int(input("Enter a number: "))
fact = 1
i = 1
while i <= n:
    fact = fact * i
    i = i + 1
print("Factorial =", fact)
```

```python
#Print all factors of a number
n = int(input("Enter a number: "))
i = 1
while i <= n:
    if n % i == 0:
        print(i)
    i = i + 1
```

```python
#Check if number is prime
n = int(input("Enter a number: "))
i = 2
prime = True
while i < n:
    if n % i == 0:
        prime = False
    i = i + 1
if prime == True:
    print("Prime Number")
else:
    print("Not Prime")
```

```python
#Guess the number game
import random
number=random.randint(1,20)
guess=None
while guess!=number:
    guess=int(input("guess number(1-20):"))
    if guess<number:
        print("Too Small")
    elif guess>number:
        print("Too Big")
    else:
        print("Correct")
```

```python
#Fibonacci series up to N terms
n = int(input("Enter number of terms: "))
a = 0
b = 1
count = 1
while count <= n:
    print(a)
    next = a + b
    a = b
    b = next
    count = count + 1
```
