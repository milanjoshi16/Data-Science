```python
#constructor
#constructor is a special method that is automatically called when an object is created
# __init__() means initialize, set the initial values of object
class student:
    def __init__(self,name,age): #Parameterised constructor
        self.name=name
        self.age=age
s1=student("bhanu",21)
print(s1.name)
print(s1.age)
```

```python
class student:
    def __init__(self,name,course,city): #Parameterised constructor
        self.name=name
        self.course=course
        self.city=city
s1=student("bhanu","DS","Jaipur")
print(s1.name)
print(s1.course)
print(s1.city)
```

```python
#getter method....used to read or access private variable
#value read
class student:
    def __init__(self,name):
        self.__name=name  #private variable with .__
    def get_name(self):
        return self.__name
s1=student("sarthak")
print(s1.get_name())
```

```python
#setter method: used to modify/update/ replace variable
class student:
    def __init__(self,name):  #(step 1) constructor runs first and saves value
        self.__name=name
    def set_name(self,name): #override vaule
        self.__name=name
    def get_name(self):
        return self.__name
s1=student("Milan")
s1.set_name("Anshu")
print(s1.get_name())
```

```python
#user input
class student:
    def __init__(self,name):  
        self.__name=name
    def set_name(self,name):
        self.__name=name
    def get_name(self):
        return self.__name
        
name = input("Enter initial name: ")
s1 = student(name)

new_name = input("Enter new name: ")
s1.set_name(new_name)

print("Updated name:", s1.get_name())
```

```python
class calculator:
#constructor
    def __init__(self):
        self.num1=0
        self.num2=0
        self.result=0
#setter method
    def set_numbers(self,a,b):
        self.num1=a
        self.num2=b
#addition method
    def add(self):
        self.result=self.num1+self.num2
#getter method
    def get_result(self):
        return self.result
#object create
calc=calculator()
calc.set_numbers(10,20)
calc.add()
print("Addition:", calc.get_result())
```

```python
#Employee Salary Management
class Employee:
    def __init__(self, salary):
        self.__salary = salary

    def set_salary(self, salary):
        self.__salary = salary

    def get_salary(self):
        return self.__salary


salary = int(input("Enter salary: "))
emp = Employee(salary)

new_salary = int(input("Enter new salary: "))
emp.set_salary(new_salary)

print("Updated Salary:", emp.get_salary())
```

```python
#Bank Account Balance
class BankAccount:
    def __init__(self, balance):
        self.__balance = balance

    def set_balance(self, balance):
        if balance >= 0:
            self.__balance = balance
        else:
            print("Invalid balance!")

    def get_balance(self):
        return self.__balance

balance = int(input("Enter initial balance: "))
account = BankAccount(balance)

print("Current Balance:", account.get_balance())

new_balance = int(input("Enter new balance: "))
account.set_balance(new_balance)

print("Final Balance:", account.get_balance())
```

# Inheritance: inherit properties


# single inheritance : one child one parent

```python
#parent class
class A:
    def show(self):
        print("Class Parent")
#child class
class B(A):
    pass
obj=B()
obj.show()
```

# Multiple inheritance: one chod multiple parent

```python
#parent class
class A:
    def showA(self):
        print("Class A")
#parent class
class B:
    def showB(self):
        print("Class B")
#child class
class C(A,B):
    pass
obj=C()
obj.showA()
obj.showB()
```
