```python
import datetime
# Get current hour
hour = datetime.datetime.now().hour
# Check time and display greeting
if 5<= hour < 12:
    print("Good Morning!")
elif 12<= hour < 18:
    print("Good Afternoon!")
elif 18<= hour < 21:
    print("Good Evening!")
else:
    print("Good Night!")
```

```python
price = float(input("Enter the price: "))
if price > 5000:
    discount_price = price - 0.20 * price
elif price > 2000:
    discount_price = price - 0.10 * price
else:
    discount_price = price - 0.05 * price
    
if discount_price > 1000:
    delivery = 0
else:
    delivery = 50
total_amount = discount_price + delivery
print("Price:", price)
print("Discounted Price:", discount_price)
print("Delivery Charge:", delivery)
print("Total Amount to be Paid:", total_amount)
```

```python
income = float(input("Enter your income (in lakhs): "))
if income <= 300000:
    tax_rate = 0
elif income <= 600000:
    tax_rate = 5
elif income <= 900000:
    tax_rate = 10
elif income <= 1200000:
    tax_rate = 15
elif income <= 1500000:
    tax_rate = 20
else:
    tax_rate = 25
tax_amount = income * tax_rate / 100
income_inhand = income - tax_amount
print("Income: Rs", income,)
print("Tax Rate:", tax_rate, "%")
print("Tax Amount: Rs", tax_amount)
print("Amount inhand: Rs", income_inhand)
```
