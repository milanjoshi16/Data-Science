```python
num=int(input("Enter number"))
if (num%3==0 & num%5==0):
    print("Number divisible by both")
elif num%5==0:
    print("Number divisible by 5")
elif num%3==0:
    print("Number divisible by 3")
elif (num%3==0 & num%5==0):
    print("Number divisible by both")
else:
    print("Number not divisible by 3&5")
```
