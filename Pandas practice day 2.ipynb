```python
# head | tail | info | shape | rename | describe 
import pandas as pd

data = {
    "Emp_ID":  [101, 102, 103, 104, 105, 106],
    "Name": ["Amit", "Riya", "Raj", "Sara", "John", "Neha"],
    "Department": ["IT", "HR", "Finance", "IT", "Sales", "HR"],
    "Salary":[50000, 45000, 60000, 55000, 48000, 52000],
    "Experience": [2, 3, 5, 4, 1, 3]
}
df = pd.DataFrame(data)
print(df)
```

```python
info = df.info()
print(info)
```

```python
shape = df.shape
print(shape)
```

```python
describe = df.describe()
print(describe)
```

```python
df.rename(columns = {"Emp_ID": "ID", "Name": "Emp_Name", "Department": "Dept"})
```

```python
print(df.tail(-2))
```
