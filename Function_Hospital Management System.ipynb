```python
patients = {}
doctors = {}
appointments = []
# Add Patient
def add_patient():
    pid = input("Enter Patient ID: ")
    if pid in patients:
        print("Patient already exists\n")
        return
    name = input("Enter Name: ")
    age = input("Enter Age: ")
    disease = input("Enter Disease: ")
    patients[pid] = {"name": name,"age": age,"disease": disease}
    print("Patient added successfully\n")
    
# View Patients
def view_patient():
    if not patients:
        print("No patient found\n")
        return
    print("\nPatient List:")
    for pid, p in patients.items():
        print(f"ID: {pid}, Name: {p['name']}, "f"Age: {p['age']}, Disease: {p['disease']}")
        print()
    
# Add Doctor
def add_doctor():
    did = input("Enter Doctor ID: ")
    if did in doctors:
        print("Doctor already exists\n")
        return
    name = input("Enter Name: ")
    specialization = input("Enter Specialization: ")
    doctors[did] = {"name": name,"specialization": specialization}
    print("Doctor added successfully\n")
    
# View Doctors
def view_doctors():
    if not doctors:
        print("No doctor found\n")
        return
    print("\nDoctor List:")
    for did, d in doctors.items():
        print(f"ID: {did}, Name: {d['name']}, "f"Specialization: {d['specialization']}")
        print()
    
# Book Appointment
def book_appointment():
    pid = input("Enter Patient ID: ")
    did = input("Enter Doctor ID: ")
    date = input("Enter Date (DD-MM-YYYY): ")
    if pid not in patients:
        print("Patient not found\n")
        return
    if did not in doctors:
        print("Doctor not found\n")
        return
    appointments.append({"pid": pid,"did": did,"date": date})
    print("Appointment booked successfully\n")
    
# View Appointments
def view_appointment():
    if not appointments:
        print("No appointment found\n")
        return
    print("\nAppointments:")
    for app in appointments:
        pid = app["pid"]
        did = app["did"]
        if pid in patients and did in doctors:
            print(f"Patient: {patients[pid]['name']} | "f"Doctor: {doctors[did]['name']} | "f"Date: {app['date']}")
            print()
    
# Search Patient
def search_patient():
    pid = input("Enter Patient ID: ")
    if pid in patients:
        p = patients[pid]
        print(f"Name: {p['name']}, "f"Age: {p['age']}, "f"Disease: {p['disease']}\n")
    else:
        print("Patient not found\n")
        
# Update Patient
def update_patient():
    pid = input("Enter Patient ID: ")
    if pid not in patients:
        print("Patient not found\n")
        return
    patients[pid]["name"] = input("Enter New Name: ")
    patients[pid]["age"] = input("Enter New Age: ")
    patients[pid]["disease"] = input("Enter New Disease: ")
    print("Patient updated successfully\n")
    
# Delete Patient
def delete_patient():
    pid = input("Enter Patient ID: ")
    if pid not in patients:
        print("Patient not found\n")
        return
    del patients[pid]
    
# Remove related appointments
global appointments
appointments = [a for a in appointments if a["pid"] != pid]
print("Patient deleted successfully\n")

# Main Menu
def menu():
    while True:
        print("===== Hospital Management System =====")
        print("1. Add Patient")
        print("2. View Patients")
        print("3. Add Doctor")
        print("4. View Doctors")
        print("5. Book Appointment")
        print("6. View Appointments")
        print("7. Search Patient")
        print("8. Update Patient")
        print("9. Delete Patient")
        print("0. Exit")
        choice = input("Enter Choice: ")
        if choice == "1":
            add_patient()
        elif choice == "2":
            view_patient()
        elif choice == "3":
            add_doctor()
        elif choice == "4":
            view_doctors()
        elif choice == "5":
            book_appointment()
        elif choice == "6":
            view_appointment()
        elif choice == "7":
            search_patient()
        elif choice == "8":
            update_patient()
        elif choice == "9":
            delete_patient()
        elif choice == "0":
            print("Program Closed")
            break
        else:
            print("Invalid Choice\n")
        
# Run Program
menu()
```
