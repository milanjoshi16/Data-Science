```python
#Global variable access
x = 10
def show():
    print(x)
show()
```

```python
#Local variable
def show():
    x = 20
    print(x)
show()
```

```python
#Global vs Local
x = 10
def show():
    x = 20
    print("Local:", x)
show()
print("Global:", x)
```

```python
#Modify global variable
x = 5
def change():
    global x
    x = 10
change()
print(x)
```

```python
#read global only
x = 100
def display():
    print(x)
display()
```

```python
#nested function
def outer():
    x = 10
    def inner():
        print(x)
    inner()
outer()
```

```python
#Nonlocal
def outer():
    x = 10
    def inner():
        nonlocal x
        x = 20
    inner()
    print(x)
outer()
```

```python
#Same Name Variable
x = 100
def show():
    x = 50
    print("Inside:", x)
show()
print("Outside:", x)
```

```python
#LEGB Rule
x = "Global"
def outer():
    x = "Enclosing"
    def inner():
        x = "Local"
        print(x)          # Local
        print(len("abc")) # Built-in
    inner()
    print(x)
outer()
print(x)
```

```python
#using globals
x = 10
def check():
    if "x" in globals():
        print("x exists")
    else:
        print("x does not exist")
check()
```

```python
#Closure
def outer(num):
    def inner(x):
        return num * x
    return inner
double = outer(2)
print(double(5))
```

```python
#Decorator
def decorator(func):
    def wrapper():
        print("Start")
        func()
        print("End")
    return wrapper

@decorator
def greet():
    print("Hello")

greet()
```

```python
#Multiple Functions Global Modify
x = 0

def add1():
    global x
    x += 3

def add2():
    global x
    x += 2

add1()
add2()

print(x)
```

```python
#locals() Function
def show():
    a = 10
    b = 20
    c = 30
    return locals()

print(show())
```

```python
#full Namespace Inteaction
x = 1
def outer():
    x = 2
    def inner():
        x = 3
        print("Inner:", x)
    inner()
    print("Outer:", x)
outer()
print("Global:", x)
```
