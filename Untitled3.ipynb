```python
def even_numbers(list1):
    for i in list1:
        if i % 2 == 0:
            print(i)
even_numbers([1, 2, 3, 4, 5, 6])
```

```python
def count_upper(text):
    count = 0
    for i in text:
        if i.isupper():
            count = count + 1
    print(count)
count_upper("PyThOn")
```

```python
def is_prime(num):
    count = 0
    for i in range(1, num + 1):
        if num % i == 0:
            count = count + 1
    if count == 2:
        print("Prime Number")
    else:
        print("Not Prime Number")
is_prime(7)
```

```python
def factorial(n):
    fact = 1
    for i in range(1, n + 1):
        fact = fact * i
    print(fact)
factorial(5)
```

```python
def reverse_string(text):
    rev = text[::-1]
    print(rev)
reverse_string("Python")
```

```python
def largest(list1):
    big = list1[0]
    for i in list1:
        if i > big:
            big = i
    print(big)
largest([10, 45, 2, 99, 23])
```

```python
def count_vowels(text):
    count = 0
    for i in text:
        if i in "aeiouAEIOU":
            count = count + 1
    print(count)
count_vowels("Education")
```

```python
def fibonacci(n):
    a = 0
    b = 1
    for i in range(n):
        print(a)
        c = a + b
        a = b
        b = c
fibonacci(5)
```

```python
def sum_digits(num):
    total = 0
    while num > 0:
        digit = num % 10
        total = total + digit
        num = num // 10
    print(total)
sum_digits(1234)
```

```python
def is_palindrome(text):
    if text == text[::-1]:
        print("Palindrome")
    else:
        print("Not Palindrome")
is_palindrome("madam")
```

```python
def remove_duplicates(list1):
    new_list = []
    for i in list1:
        if i not in new_list:
            new_list.append(i)
    print(new_list)
remove_duplicates([1, 2, 2, 3, 4, 4, 5])
```

```python
def calculator(a, b, op):
    if op == "+":
        print(a + b)
    elif op == "-":
        print(a - b)
    elif op == "*":
        print(a * b)
    elif op == "/":
        print(a / b)
calculator(10, 5, "+")
calculator(10, 5, "-")
calculator(10, 5, "*")
calculator(10, 5, "/")
```

```python
def print_even(n):
    if n < 2:
        return 0
    if n % 2 == 0:
        print(n)
        print_even(n - 2)
    else:
        print_even(n - 1)
print_even(10)
```
