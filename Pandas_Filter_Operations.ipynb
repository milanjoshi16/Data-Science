```python
import pandas as pd
data = "https://raw.githubusercontent.com/rajendra0968jangid/Ds-Arya/main/student-data.json"
df = pd.read_json(data)
print(df)
```

```python
#single column
print(df["maths"])
```

```python
filter_data = df["maths"]>50
#print(filter_data)---- returns value in true false
#print(df[filter_data]) -----print the filtered data itself
print(df[df["maths"]>50]) #----- entire step in one statement
```

```python
#english < 60
print(df[df["english"]<60])
```

```python
print(df[df["physics"]==78])
```

```python
print(df[df["physics"]!=78])
```

```python
print(df[df["name"]  == "Sunder"])
```

```python
print(df[ (df["maths"]>60) & (df["english"]<60)])
```

```python
print(df[ (df["gender"]=="male") & (df["maths"]>60) ])
```

```python
print(df[ (df["gender"]=="male") | (df["maths"]>60) ])
```

```python
print(df[ (df["maths"]>60) & (df["english"]<60)]["physics"])
```

```python
print(df[ (df["maths"]>60) & (df["english"]<60)][["physics","maths"]])
```
