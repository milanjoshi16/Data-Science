```python
#Unpack a tuple
t = (10, 20, 30)
a, b, c = t
print(a)
print(b)
print(c)
```

```python
#Unpack a list
lst = [1, 2, 3]
a, b, c = lst
print(a, b, c)
```

```python
#Extended unpacking
a, *b = [1, 2, 3, 4, 5]
print(a)
print(b)
```

#First, middle and last elements
lst = [10, 20, 30, 40]
first, *middle, last = lst
print(first)
print(middle)
print(last)

```python
#Swap two variables
a = 5
b = 10
a, b = b, a
print(a)
print(b)
```

```python
#Function return unpacking
def fun():
    return 1, 2, 3
a, b, c = fun()
print(a, b, c)
```

# zip()

```python
#Combine two lists
names = ["A", "B", "C"]
marks = [90, 80, 70]
result = zip(names, marks)
print(list(result))
```

```python
#Convert zip object into list of tuples
a = [1, 2, 3]
b = ['x', 'y', 'z']
result = list(zip(a, b))
print(result)
```

```python
#Create dictionary using zip
names = ["A", "B", "C"]
marks = [90, 80, 70]
d = dict(zip(names, marks))
print(d)
```

```python
#Unequal length lists
a = [1, 2, 3]
b = ['a', 'b']
print(list(zip(a, b)))

#zip() stops at the shortest list
```

```python
#Use zip with a loop
names = ["A", "B", "C"]
marks = [90, 80, 70]
for name, mark in zip(names, marks):
    print(name, mark)
```

```python
#Unzip data
pairs = [(1, 'a'), (2, 'b')]
numbers, letters = zip(*pairs)
print(numbers)
print(letters)
```

# Enumerate()

```python
#Print elements with index
lst = ['a', 'b', 'c']
for i, value in enumerate(lst):
    print(i, value)
```

```python
#Start index from 1
lst = ['a', 'b', 'c']
for i, value in enumerate(lst, start=1):
    print(i, value)
```

```python
#Enumerate a string
text = "Python"
for i, ch in enumerate(text):
    print(i, ch)
```

```python
#Print elements at even indices
lst = [10, 20, 30, 40, 50]
for i, value in enumerate(lst):
    if i % 2 == 0:
        print(value)
```

```python
#Print elements where index > 2
lst = [10, 20, 30, 40, 50]
for i, value in enumerate(lst):
    if i > 2:
        print(value)
```

```python
#Enumerate dictionary keys
d = {'a': 1, 'b': 2, 'c': 3}
for i, key in enumerate(d):
    print(i, key)
```

# List Comprehension

```python
#Squares from 1 to 10
squares = [i**2 for i in range(1, 11)]
print(squares)
```

```python
#Even numbers from 1 to 50
evens = [i for i in range(1, 51) if i % 2 == 0]
print(evens)
```

```python
#String to list of characters
text = "Python"
chars = [ch for ch in text]
print(chars)
```

```python
#Filter odd numbers
lst = [1, 2, 3, 4, 5, 6]
odd = [i for i in lst if i % 2 != 0]
print(odd)
```

```python
#Flatten a 2D list
lst = [[1, 2], [3, 4], [5, 6]]
flat = [j for i in lst for j in i]
print(flat)
```

```python
#If-else in list comprehension
lst = [1, 2, 3, 4, 5]
result = ["Even" if i % 2 == 0 else "Odd" for i in lst]
print(result)
```

# Dictionary Comprehension

```python
#Squares dictionary
d = {i: i**2 for i in range(1, 6)}
print(d)
```

```python
#Dictionary from two lists
keys = ["a", "b", "c"]
values = [1, 2, 3]
d = {k: v for k, v in zip(keys, values)}
print(d)
```

```python
#Filter dictionary (even values)
d = {'a': 1, 'b': 2, 'c': 4, 'd': 5}
result = {k: v for k, v in d.items() if v % 2 == 0}
print(result)
```

```python
#Swap keys and values
d = {'a': 1, 'b': 2, 'c': 3}
swapped = {v: k for k, v in d.items()}
print(swapped)
```
