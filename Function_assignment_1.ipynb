```python
# max of 2 numbers
def greater(a,b):
    if a>b:
        print("Greater Number is",a)
    else:
        print("Greater Number is",b)
greater(35,38)
```

```python
#sum of 1 to n
def sum_n(n):
    total=0
    for i in range(1,n+1):
        total = total + i
    print("Sum is", total)
    
sum_n(3)
```

```python
#positive negative or zero
def check_num(n):
    if n > 0:
        print("Number is positive")
    elif n < 0:
        print("Number is negative")
    else:
        print("Number is zero")

check_num(-10)
```

```python
#count vowels in a string
def count_v(text):
    count = 0
    for ch in text:
        if ch in "aeiouAEIOU":
            count=count+1
    print(count)

count_v("Uilan")
```

```python
#reverse a string
def reverse_string(text):
    print(text[::-1])

reverse_string("Python")
```

```python
#factorial
def factorial(n):
    fact = 1
    for i in range (1,n+1):
        fact= fact * i
    print(fact)

factorial(5)
```

```python
#Prime Number
def prime(num):
    count = 0
    for i in range(1, num + 1):
        if num % i == 0:
            count += 1
    if count == 2:
        print("Prime")
    else:
        print("Not Prime")

prime(7)
```

```python
#length of list without len()
def list_length(lst):
    count = 0
    for i in lst:
        count += 1
    print(count)

list_length([1, 2, 3, 4, 5])
```

```python
#table
def table(num):
    for i in range(1, 11):
        print(num, "x", i, "=", num * i)

table(5)
```

```python
#take list and return largest number
def largest(lst):
    large = lst[0]
    for i in lst:
        if i > large:
            large = i
    return large

print(largest([15, 49, 18, 90, 25]))
```
