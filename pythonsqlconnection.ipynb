```python
import pymysql as sql
connection = sql.connect(host="localhost", port = 3306, user = "root", password="12345", database="python")
cursor=connection.cursor()
student_id=int(input("Enter student id:"))
name=input("Enter name:")
age=int(input("Enter age:"))
gender=input("Enter gender")
marks=int(input("Enter marks"))
query=""" insert into student(student_id,name,age,gender,marks) values(%s,%s,%s,%s,%s)"""
cursor.execute(query,(student_id,name,age,gender,marks))
connection.commit()
print("Record inserted successfully")
```
