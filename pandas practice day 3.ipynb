```python
import pandas as pd

data = {
    "Emp_ID": [201, 202, 203, 204, 205, 206],
    "Name": ["Arjun", "Priya", "Karan", "Meena", "David", "Anjali"],
    "Department": ["Marketing", "IT", "HR", "Finance", "Sales", "IT"],
    "Salary": [58000, 62000, 47000, 65000, 51000, 59000],
    "Experience": [3, 5, 2, 6, 2, 4]
}
df = pd.DataFrame(data)
print(df)
```

```python
#select column
print(df["Emp_ID"])
```

```python
#add new column
df["Leaves"]= [10,15,12,9,11,16]
print(df)
```

```python
#delete column
df.drop("Levels", axis = 1, inplace=True)
print(df)
```

```python
#type
type(df[["Salary","Experience"]])
```

```python
#multiple columns to delete
df.drop(["Name","Department"], axis=1, inplace = True) # data not deleted until use of inplace=True
print(df)
```

```python
#select multiple columns
df[["Emp_ID","Salary"]]
```

```python
#add new columns
df["Bonus"]=df["Salary"]
print(df)
```

```python
#add new column-> increment only
df["Increment"]=df["Salary"] + 0.10
print(df)
```

```python
import pandas as pd
url="https://raw.githubusercontent.com/rajendra0968jangid/Ds-Arya/main/student-data.json"
df= pd.read_json(url)
print(df)
```

```python
df['Subject_Total'] = df[['physics', 'maths', 'english']].sum(axis=1)
print(df)
```

```python
df['Subject_Average'] = df["Subject_Total"]/3
print(df)
```
