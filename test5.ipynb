```python
# Area of a Rectangle
def calculate_area(length, width):
    return length * width

print(calculate_area(6, 4))
```

```python
#Find Larger Number
def find_max(a, b):
    if a > b:
        return a
    else:
        return b
print(find_max(10, 20))
```

```python
# Count vowels in a String
def count_vowels(text):
    count = 0
    for ch in text:
        if ch.lower() in "aeiou" or ch.upper() in "AEIOU":
            count += 1
    return count
print(count_vowels("Hello World A"))
```

```python
#Prine Number Check
def is_prime(n):
    if n <= 1:
        return False
    for i in range(2, n):
        if n % i == 0:
            return False
    return True
print(is_prime(7))
```

```python
#Increase Global Variable by 10
num = 50
def increase():
    global num
    num = num + 10
increase()
print(num)
```

```python
#Lambda Function
square = lambda x: x * x
print(square(5))
```

```python
#map() for Squares
numbers = [1, 2, 3, 4, 5]
result = list(map(lambda x: x*x, numbers))
print(result)
```

```python
#filter() for Odd Numbers
numbers = [10, 15, 20, 25, 30, 35]
result = list(filter(lambda x: x % 2 != 0, numbers))
print(result)
```

```python
#Sort Tuples by Age
people = [("John", 25), ("Alice", 20), ("Bob", 30)]
result = sorted(lambda x: key=x[1], people)
print(result)
```

```python
#Recursive Function: Print N to 1
def print_numbers(n):
    if n == 0:
        return 
    print(n)
    print_numbers(n-1)
print_numbers(5)
```

```python
#Recursive Sum of First N Natural Numbers
def sum_n(n):
    if n == 0:
        return 0
    return n+sum_n(n-1)
print(sum_n(5))
```

```python
#Recursive Count Digits
def count_digits(n):
    if n ==0:
        return 0
    return 1+count_digits(n // 10)
print(count_digits(12345))
```
