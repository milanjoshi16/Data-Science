```python
import random
player=0
while player < 100:
    a=input("Press 1 to roll the dice ")
    dice=random.randint(1, 6)
    print("You got:", dice)
    player=player + dice
    if (player==18 or player==25 or player==34 or player==45 or player==52 or player==77 or player==91 or player==97):
        print("Ladder!")
        inc=random.randint(1,100)
        if(player < inc):
            print("increment:", inc)
            player=player + random.inc
    elif (player==5 or player ==13 or player==27 or player==36 or player==49 or player== 63 or player==87 or player==91 or player==99):
        print("Snake!")
        dec=random.randint(1, player)
        if(player > dec):
            print("Decrement:", dec)
            player= player - dec
    print("Current Position:", player)
if player == 100:
    print("You Win!!")
```
