```python
#Count the Number of Each Vowel in a String
text = input("Enter a string: ")
v=['a','e','i','o','u']
count=0
for ch in text:
    if ch in v:
        count+=1
print (count)
```

```python
#Count Words, Characters and Spaces in a String
text = input("Enter a string: ")
words = 1
spaces = 0
characters = 0
for ch in text:
    characters = characters + 1
    if ch == " ":
        spaces = spaces + 1
        words = words + 1
print("Words =", words)
print("Characters =", characters)
print("Spaces =", spaces)
```

```python
#Remove Punctuation from a String
text = input("Enter a string: ")
result = ""
punctuation = "!@#$%^&*(),.?\":{}|<>"
for ch in text:
    if ch not in punctuation:
        result = result + ch
print("After removing punctuation:")
print(result)
```

```python
#Take Matrix Input and Store in Nested List
rows = int(input("Enter number of rows: "))
cols = int(input("Enter number of columns: "))
matrix = []
for i in range(rows):
    row = []
    for j in range(cols):
        value = int(input("Enter element: "))
        row.append(value)
    matrix.append(row)
print("Matrix =")
for row in matrix:
    print(row)
```

```python
#Add Two Matrices
A = [[1, 2], [3, 4]]
B = [[5, 6], [7, 8]]
result = [[0, 0], [0, 0]]
for i in range(2):
    for j in range(2):
        result[i][j] = A[i][j] + B[i][j]
print("Addition =")
for row in result:
    print(row)
```

```python
#Multiply Two Matrices
A = [[1, 2], [3, 4]]
B = [[5, 6], [7, 8]]
result = [[0, 0], [0, 0]]
for i in range(2):
    for j in range(2):
        for k in range(2):
            result[i][j] = result[i][j] + A[i][k] * B[k][j]
print("Multiplication =")
for row in result:
    print(row)
```
