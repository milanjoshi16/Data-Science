```python
#Banking withdrawal System
balance = 10000
amt = int(input("Enter withdrawal amount: "))
if amt < 0:
    print("Invalid amount")
elif amt > balance:
    print("Insufficient funds")
elif amt >= 5000:
    print("OTP Required")
else:
    print("Withdrawal Successful")
```

```python
#Online Exam Result Analyzer
eng = int(input("Enter marks of English: "))
hindi= int(input("Enter marks of Hindi: "))
maths= int(input("Enter marks of Maths: "))
science= int(input("Enter marks of Science: "))
sst= int(input("Enter marks of SSt: "))
average = (eng + hindi + maths + science + sst) / 5
if eng < 33 or hindi < 33 or maths < 33 or science < 33 or sst < 33:
    print("Fail")
elif average >= 75:
    print("Distinction")
elif average >= 60:
    print("First Division")
elif average >= 50:
    print("Second Division")
else:
    print("Third Division")
```

```python
#Smart Traffic Light System
speed = int(input("Enter vehicle speed: "))
signal = input("Enter signal color: ")

if speed < 0:
    print("Invalid speed")

elif signal == "red" and speed > 0:
    print("Violation")

elif signal == "yellow" and speed > 20:
    print("Warning")

elif signal == "green":
    print("Allowed")

else:
    print("Safe Driving")
```

```python
#ATM PIN Attempts
correct_pin = "1234"
pin = input("Enter PIN: ")
if pin == correct_pin:
    print("Access Granted")
else:
    print("Wrong PIN")
    pin = input("Enter PIN again: ")
    if pin == correct_pin:
        print("Access Granted")
    else:
        print("Wrong PIN")
        pin = input("Enter PIN third time: ")
        if pin == correct_pin:
            print("Access Granted")
        else:
            print("Card Blocked")
```

```python
#Loan Eligibility Checker
age = int(input("Enter age: "))
salary = int(input("Enter salary: "))
score = int(input("Enter credit score: "))
if age < 21 or age > 60:
    print("Loan Rejected")
elif score < 600:
    print("Loan Rejected")
elif salary < 25000:
    print("Loan Rejected")
elif score > 750 and salary > 50000:
    print("Premium Approval")
else:
    print("Normal Approval")
```

```python
#Login Security System
correct_username = "admin"
correct_password = "1234"
correct_captcha = "AB12"
username = input("Enter username: ")
password = input("Enter password: ")
captcha = input("Enter captcha: ")
if username != correct_username:
    print("Invalid User")
elif password != correct_password:
    print("Incorrect Password")
elif captcha != correct_captcha:
    print("Verification Failed")
else:
    print("Login Success")
```

```python
#Shopping Coupon Validator
code = input("Enter coupon code: ")
value = int(input("Enter cart value: "))
if code == "SAVE10" and value >= 500:
    print("10% OFF")
elif code == "SAVE20" and value >= 2000:
    print("20% OFF")
elif code == "SAVE50" and value >= 5000:
    print("50% OFF")
elif code != "SAVE10" and code != "SAVE20" and code != "SAVE50":
    print("Invalid Coupon")
else:
    print("Conditions Not Met")
```

```python
#Hospital Priority System
age = int(input("Enter age: "))
severity = int(input("Enter severity level: "))
if severity >= 4:
    print("Immediate")
elif severity == 3:
    print("Urgent")
elif severity <= 2 and age > 60:
    print("Priority")
else:
    print("Normal")
```
