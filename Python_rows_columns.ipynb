```python
import numpy as np
 
students = np.array([
    [101, 78, 85, 92, 88], # id,maths,english,computer,science
    [102, 65, 72, 70, 68],
    [103, 90, 95, 91, 94],
    [104, 45, 55, 60, 50],
    [105, 82, 79, 85, 80],
    [106, 55, 60, 58, 62],
    [107, 88, 92, 86, 90],
    [108, 72, 68, 75, 70],
    [109, 95, 91, 96, 94],
    [110, 40, 48, 45, 42]
])
 
print(students)
```

```python
#shape,size,dimention
print("Shape:", students.shape) #10 rows, 5 columns
print("Size:", students.size) #no. of elements
print("Dimentions", students.ndim) #2d
```

```python
# reverse data set (row_start: row_stop: row_step, )
print(students[::-1]) # last rows becomes first
print("Reverse skip by 2")
print(students[::-2])
```

```python
# reverse subject order (column change)
print(students[:,1:][:,::-1])
```

```python
#reverse subject order/ alternative way
f= students[:,1:]
print(f[:,::-1])
```

```python
#total | sum of marks of all subjects and particular student
marks = students[:,1:]
total= np.sum(marks, axis = 1) #by default row but we need columns
print(total)

# alternatively
for i in range(len(marks)):
    count=0
    for j in range(4):
        count += marks[i][j]
    print(f"students {i} marks {count}")
    
```
