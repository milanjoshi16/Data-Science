```python
#Positive, Negative and Zero
num = int(input("Enter a number: "))
if num > 0:
    print("It's a Positive Number")
elif num < 0:
    print("It's a Negative Number")
else:
    print("Zero")
if num % 2 == 0:
    print("It's an Even Number")
else:
    print("It's an Odd Number")
```

```python
#Biggest Among Three Numbers
num1 = int(input("Enter first number: "))
num2 = int(input("Enter second number: "))
num3 = int(input("Enter third number: "))
if num1 > num2 and num1 > num3:
    print("Greatest number is:", num1)
elif num2 > num1 and num2 > num3:
    print("Greatest number is:", num2)
else:
    print("Greatest number is:", num3)
```

```python
#Electricity Bill Calculator
e_units = int(input("Enter electricity units: "))
if e_units <= 100:
    bill = e_units * 5
elif e_units <= 200:
    phase2=(e_units-100)*7
    bill = (100 * 5) + phase2
else:
    bill = (100 * 5) + (100 * 7) + ((e_units - 200) * 10)
print("Total Bill =", bill)
```

```python
#Strong Password Checker
password = input("Enter password: ")
if len(password) < 8:
    print("Weak Password")
elif password.islower():
    print("Weak Password")
elif password.isalpha():
    print("Weak Password")
else:
    print("Strong Password")
```

```python
#Triangle Validity Checker
angle1 = int(input("Enter first angle: "))
angle2 = int(input("Enter second angle: "))
angle3 = int(input("Enter third angle: "))
if angle1 > 0 and angle2 > 0 and angle3 > 0 and (angle1 + angle2 + angle3 == 180):
    print("Valid Triangle")
    if angle1 > 90 or angle2 > 90 or angle3 > 90:
        print("Obtuse Angle Triangle")
    elif angle1 == 90 or angle2 == 90 or angle3 == 90:
        print("Right Angle Triangle")
    else:
        print("Acute Angle Triangle")
else:
    print("Invalid Triangle")
```

```python
#Income Tax Calculator
income = float(input("Enter your income: "))
if income < 250000:
    tax = 0
elif income <= 500000:
    tax = income * 0.05
elif income <= 1000000:
    tax = income * 0.20
else:
    tax = income * 0.30
print("Total Tax =", tax)
```

```python
#Rock Paper Scissors Game
player1 = input("User1 Enter Rock/Paper/Scissors: ")
player2 = input("User2 Enter Rock/Paper/Scissors: ")
if player1 == player2:
    print("Match Draw")
elif player1 == "Rock" and player2 == "Scissors":
    print("Player1 Wins")
elif player1 == "Paper" and player2 == "Rock":
    print("Player1 Wins")
elif player1 == "Scissors" and player2 == "Paper":
    print("Player1 Wins")
else:
    print("Player2 Wins")
```

```python
#Menu Driven Food Order System
print("1. Pizza = 200")
print("2. Burger = 120")
print("3. Pasta = 150")
print("4. Momos = 100")
choice = int(input("Enter your choice: "))
qty = int(input("Enter quantity: "))
if choice == 1:
    bill = 200 * qty
elif choice == 2:
    bill = 120 * qty
elif choice == 3:
    bill = 150 * qty
elif choice == 4:
    bill = 100 * qty
else:
    bill = 0
    print("Invalid Choice")
if bill > 1000:
    discount = bill * 0.10
    bill = bill - discount
    print("10% Discount Applied")
print("Total Bill =", bill)
```
